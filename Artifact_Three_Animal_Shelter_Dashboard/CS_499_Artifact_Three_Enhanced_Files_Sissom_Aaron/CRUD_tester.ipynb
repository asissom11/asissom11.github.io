# Test Script for db_function_tester.py

Test environment for CRUD operations (Create, Read, Update, Delete) that live in the `database_functions.py` file.

## User Authentication

- Import CRUD modules from test file
- Instantiate a "user" object

```python
from db_function_tester import AnimalShelter
user = AnimalShelter('aacuser', 'CS340', 'admin')
```

## Testing CREATE method

- Create a record to be inserted into the database collection "AAC.animals"

```python
record_one = [{'animal_type':'Dog', 'date_time_group':'2024-10-01', 'animal_name':'Taco'}]
user.create(record_one)
```

## Testing READ method

- Query the database to return the specified record(s)
- Comparison operator `$in` used to filter by specified parameter. In this example there is only 'Dog', but the principle is still valid.

```python
record_two = {'animal_type': {'$in': ['Dog']}}
user.read(record_two)
```

## Testing UPDATE method

- Query the database to return items to update, then update them and replace in the database.
- Using `$ne` is the same as the `!=` inequality; using `$set` replaces the value of a field with a specified value

```python
record_three = ({'animal_type': 'Dog', 'animal_name': {'$ne': 'Bailey'}}, {'$set': {'animal_name': 'Bailey', 'update': 'true'}})
user.update(record_three)
```

## Testing DELETE method

- Query the database to return a record for removal based on a given pattern

```python
record_four = {'animal_name': {'$regex': '^Bailey'}}
user.delete(record_four)
```
